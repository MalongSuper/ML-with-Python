## Logistic Regression

**Logistic Regression** is a supervised machine learning algorithm used for classification problems, where the goal is to predict a discrete category rather than a continuous value. Despite its name, it is a classification model that estimates the probability of an observation belonging to a particular class by applying a logistic (sigmoid) function to a linear combination of input features. 

The sigmoid function:

$$𝒇(𝒙) = \frac{𝟏}{𝟏+𝒆^{−𝒙}}$$


Helps transform any real-valued output into a value between 0 and 1, which can be interpreted as a probability. For binary classification, predictions are typically assigned to one class if the probability exceeds a threshold (commonly 0.5) and to the other class otherwise.

Logistic Regression is widely used because it is simple, computationally efficient, easy to interpret, and often serves as a strong baseline model for tasks such as spam detection, medical diagnosis, customer churn prediction, and sentiment analysis.


```python
import pandas as pd
import numpy as np

# Read the data
data = {"Study_Hours": [2, 3, 4, 5, 6, 7, 8, 1, 9, 10, 3,6],
        "Mock_Test_Score": [4.5, 5.0, 5.5, 6.0, 6.5, 7.0, 7.5, 4.0, 8.0, 8.5, 4.8, 6.8],
        "Extra_Class": [0, 0, 1, 0, 1, 1, 0, 0, 1, 1, 1, 0],
        "Label": [0, 0, 1, 0, 1, 1, 1, 0, 1, 1, 0, 1]}

df = pd.DataFrame(data)

# Coefficients from Iteration 1 (scaled)
B0, B1, B2, B3 = 0.1, 1.8, 1.275, 0.2

# Compute z_i
df["z_i"] = B0 + B1*df["Study_Hours"] + B2*df["Mock_Test_Score"] + B3*df["Extra_Class"]
# Sigmoid p_i
df["p_i"] = 1 / (1 + np.exp(-df["z_i"]))
# Error
df["p_i - y_i"] = df["p_i"] - df["Label"]

print(df[["z_i", "p_i", "p_i - y_i"]])
```

```python
# Gradients
grad_B0 = df["p_i - y_i"].sum()
grad_B1 = (df["p_i - y_i"] * df["Study_Hours"]).sum()
grad_B2 = (df["p_i - y_i"] * df["Mock_Test_Score"]).sum()
grad_B3 = (df["p_i - y_i"] * df["Extra_Class"]).sum()

print("Gradients:")
print(grad_B0, grad_B1, grad_B2, grad_B3)
```

```python
alpha = 0.1

B0_new = B0 - alpha * grad_B0
B1_new = B1 - alpha * grad_B1
B2_new = B2 - alpha * grad_B2
B3_new = B3 - alpha * grad_B3

print("Updated Coefficients:")
print(f"B0 = {B0_new}, B1 = {B1_new}, B2 = {B2_new}, B3 = {B3_new}")
```
