## Gradient Boosting

Gradient Boosting is an ensemble learning technique that builds a strong predictive model by combining multiple weak learners, typically decision trees. The idea is to sequentially add models that correct the errors of the previous models, resulting in a powerful overall model. Each new model is trained on the residuals (errors) of the previous model, and the final prediction is made by summing the predictions of all the individual models. Gradient Boosting is widely used for both classification and regression tasks due to its high accuracy and ability to handle complex data.

It involves a stump (a decision tree with a single split) as the weak learner, and the algorithm iteratively adds stumps to the ensemble, minimizing the loss function at each step. The learning rate is a hyperparameter that controls how much each new model contributes to the overall prediction, allowing for better generalization and preventing overfitting.

Then the first decision tree is built using the original dataset, and the residuals (the difference between the actual target values and the predicted values) are calculated. The next decision tree is then trained on these residuals, and this process continues until a specified number of trees is reached or the residuals are sufficiently small.

So the regression trees are added sequentially, and each tree is trained to predict the residuals of the previous tree, which helps to improve the overall model's performance. The final prediction is made by summing the predictions of all the individual trees in the ensemble.

The residuals are calculated as follows:

$$r_i = y_i - \hat{y}_i$$

Where:
- $r_i$ is the residual for the $i$-th data point.
- $y_i$ is the actual target value for the $i$-th data point.
- $\hat{y}_i$ is the predicted value for the $i$-th data point from the current model.

The following formula is used to update the predictions at each iteration:


$$\hat{y}_i^{(new)} = \hat{y}_i^{(old)} + \eta \cdot r_i^{(pred)}$$

Where:

- $\hat{y}_i^{(new)}$ is the updated prediction for the $i$-th data point.
- $\hat{y}_i^{(old)}$ is the previous prediction for the $i$-th data point.
- $\eta$ is the learning rate, which controls the contribution of each new tree to the overall prediction.
- $r_i^{(pred)}$ is the predicted residual for the $i$-th data point from the current tree.

```python
import pandas as pd
 
data = {'Age': ['≤30', '≤30', '31…40', '>40', '>40', '>40', '31…40', '≤30', '≤30', '>40', '≤30', '31…40', '31…40', '>40'], 
        'Income': ['high', 'high', 'high', 'medium', 'low', 'low', 'low', 'medium', 'low', 'medium', 'medium', 'medium', 'high', 'medium'], 
        'Student': ['no', 'no', 'no', 'no', 'yes', 'yes', 'yes', 'no', 'yes', 'yes', 'yes', 'no', 'yes', 'no'], 
        'Credit_Rating': ['Fair', 'Excellent', 'Fair', 'Fair', 'Fair', 'Excellent', 'Excellent', 'Fair', 'Fair', 'Fair', 'Excellent', 'Excellent', 'Fair', 'Excellent'],
        'Buy_XBOX': ['no', 'no', 'yes', 'yes', 'yes', 'no', 'yes', 'no', 'yes', 'yes', 'yes', 'yes', 'yes', 'no']}

df = pd.DataFrame(data)

# We use Label Encoding for demonstration only
for i in df.columns:
  unique_columns = df[i].unique()
  df[i] = df[i].map({unique_columns[value]: value for value in range(len(unique_columns))})

# As the first prediction for GB, use probability of the common class
df['Predicted y_0'] = max(df['Buy_XBOX'].value_counts(normalize=True).unique())
print(df[['Buy_XBOX', 'Predicted y_0']])
```

```python
df['Residual r_0'] = df['Buy_XBOX'] - df['Predicted y_0']
print(df[['Buy_XBOX', 'Predicted y_0', 'Residual r_0']])
```

```python
import numpy as np

# Generate random prediction of the residual
np.random.seed(42)
df['Predict Residual r_0'] = np.random.rand(len(df))
# Update the prediction
learning_rate = 0.1
df['Predicted y_1'] = df['Predicted y_0'] + learning_rate *  df['Predict Residual r_0']
print(df[['Predicted y_0', 'Residual r_0', 'Predict Residual r_0', 'Predicted y_1']])
```

```python
df['Residual r_1'] = df['Predicted y_1'] - df['Buy_XBOX']
print(df[['Buy_XBOX', 'Predicted y_1', 'Residual r_1']])
```

```python
df['Predict Residual_r1'] = np.random.rand(len(df))
df['Predicted y_2'] = df['Predicted y_1'] + learning_rate *  df['Predict Residual_r1']
print(df[['Predicted y_1', 'Residual r_1', 'Predict Residual_r1', 'Predicted y_2']])
```

The sigmoid function is often used in classification tasks to convert the output of the model into a probability between 0 and 1. The formula for the sigmoid function is:

$$\sigma(x) = \frac{1}{1 + e^{-x}}$$

Where:
- $\sigma(x)$ is the output of the sigmoid function, which represents the probability of the positive class.
- $x$ is the input to the sigmoid function, which is typically the output of the model before applying the sigmoid transformation. 

```python
# Use the sigmoid function to convert the predicted y_2 to probability
def sigmoid(x):
  return 1 / (1 + np.exp(-x))

df['Predicted Probability y_2'] = sigmoid(df['Predicted y_2'])
df['Predicted Buy_XBOX'] = np.where(df['Predicted Probability y_2'] >= np.mean(df['Predicted Probability y_2']), 1, 0)
df['Actual Buy_XBOX'] = df['Buy_XBOX']
print(df[['Predicted Probability y_2', 'Predicted Buy_XBOX', 'Actual Buy_XBOX']])
# Accuracy
print(f"Accuracy: {np.mean(df['Predicted Buy_XBOX'] == df['Actual Buy_XBOX'])}")
```
