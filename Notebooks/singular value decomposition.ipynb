## Singular Value Decomposition

Singular Value Decomposition (SVD) is a powerful matrix factorization technique used in various applications, including recommender systems. It decomposes a matrix into three other matrices, capturing the underlying structure of the data. In the context of recommender systems, SVD can be used to predict user preferences and make recommendations based on the latent factors derived from the user-item interaction matrix.

Formula:
$$A = U \Sigma V^T$$

Where:
- $A$ is the original user-item interaction matrix.

- $U$ is the matrix of left singular vectors (user factors).

- $V^T$ is the matrix of right singular vectors (item factors).

- $\Sigma$ is the diagonal matrix of singular values (importance of each factor).

```python
import numpy as np

A = np.array([[5, 4, 2, 3],
              [3, 2, 3, 3],
              [4, 4, 3, 1],
              [1, 2, 3, 1],
              [1, 1, 5, 1]])

ATA = A.T @ A
print(f"A^T * A:\n", ATA)
```

```python
# Eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(ATA)
print("Eigenvalues:\n", eigenvalues)
print("Eigenvectors:\n", eigenvectors)
```

```python
V = eigenvectors
print("V:\n", V)
```

```python
sigma = np.sqrt(eigenvalues)
U = []
for i in range(len(sigma)):
    if sigma[i] > 1e-10:
        ui = (A @ V[:, i]) / sigma[i]
        U.append(ui)

U = np.array(U).T
print("U:\n", U)
```

```python
sigma = np.sqrt(eigenvalues)
Sigma = np.diag(sigma)
print("Sigma:\n", Sigma)
```

```python
A_reconstructed = U @ Sigma @ V.T
print("A_reconstructed:\n", A_reconstructed)
```
