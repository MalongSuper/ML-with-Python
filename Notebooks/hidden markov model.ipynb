## Hidden Markov Models

Hidden Markov Models (HMMs) are a statistical model used to represent systems that are assumed to be a Markov process with hidden states. They are widely used in various fields such as speech recognition, natural language processing, and bioinformatics.

Key components of an HMM include:
1. **States**: A finite set of hidden states that the system can be in.

2. **Observations**: A finite set of observable events that are generated by the hidden states.

3. **Transition Probabilities**: The probabilities of transitioning from one hidden state to another.

4. **Emission Probabilities**: The probabilities of an observation being generated from a hidden state.

5. **Initial State Probabilities**: The probabilities of the system starting in each hidden state.

```python
import numpy as np

# Transition matrix
A = np.array([[0.4, 0.4, 0.2],
              [0.1, 0.8, 0.1],
              [0.3, 0.2, 0.5]])

# Initial state: Happy
p = np.array([0.0, 1.0, 0.0])

# Evolve for 10 steps
for i in range(10):
    p = np.dot(p, A)
    print(p)
```

The forward algorithm is a dynamic programming algorithm used to compute the probability of a sequence of observed events given a hidden Markov model. It is used to calculate the likelihood of a sequence of observations, which can be useful for tasks such as speech recognition and natural language processing.

```python
import numpy as np

def forward_algorithm(obs, hidden_states, transition_matrix, emission_matrix,
                     init=np.array([1/3, 1/3, 1/3])):

    T = len(obs)      # number of time steps
    N = len(hidden_states)  # number of states

    alpha = np.zeros((T, N))

    # Initialization (t = 1)
    O0 = obs_map[obs[0]]
    alpha[0, :] = init * emission_matrix[:, O0]
    print("alpha[0]:", alpha[0])

    # Recursion
    for t in range(1, T):
        Ot = obs_map[obs[t]]
        for j in range(N):
            alpha[t, j] = (np.sum(alpha[t-1, :] * transition_matrix[:, j])
                           * emission_matrix[j, Ot])
        print(f"alpha[{t}]:", alpha[t])

    # Final probability
    P_O = np.sum(alpha[T-1, :])
    print("\nProbability of the observation sequence:", P_O)
```

```python
observations = ["Walk", "Shop", "Clean", "Walk", "Walk",
                "Shop", "Clean", "Walk", "Clean", "Shop"]

hidden_states = ["Sad", "Happy", "Angry"]

obs_map = {"Walk": 0, "Clean": 1, "Shop": 2}

A = np.array([[0.4, 0.4, 0.2],
              [0.1, 0.8, 0.1],
              [0.3, 0.2, 0.5]])

B = np.array([[0.1, 0.2, 0.7],
              [0.6, 0.3, 0.1],
              [0.2, 0.6, 0.2]])

pi = np.array([1/3, 1/3, 1/3])

forward_algorithm(observations, hidden_states, A, B, pi)
```
